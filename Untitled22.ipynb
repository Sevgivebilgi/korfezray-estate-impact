```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. Kocaeli Excel veri setini yüklüyoruz
file_path = 'KOCAELI.xlsx'
df = pd.read_excel(file_path, sheet_name='Sayfa1')

# 2. 'ZAMAN' sütunu hariç tüm konut satış değişkenlerini seçiyoruz
sales_cols = [col for col in df.columns if col != 'ZAMAN']

# 3. Doğal logaritma (ln) dönüşümü ve Jarque-Bera analizi
jb_log_results = []
for col in sales_cols:
    # Verinin doğal logaritması
    data_log = np.log(df[col])

    mean_val = data_log.mean()
    std_val = data_log.std()
    median_val = data_log.median()
    min_val = data_log.min()
    max_val = data_log.max()

    # Jarque-Bera Normallik Testi
    jb_stat, jb_p = stats.jarque_bera(data_log)

    jb_log_results.append({
        'Değişken': f'ln_{col}',
        'Ortalama': round(mean_val, 4),
        'Std. Sapma': round(std_val, 4),
        'Medyan': round(median_val, 4),
        'Min': round(min_val, 4),
        'Max': round(max_val, 4),
        'JB İstatistiği': round(jb_stat, 4),
        'JB p-değeri': round(jb_p, 4)
    })

# DataFrame'e dönüştürüp ekrana yazdırıyoruz
df_jb_log = pd.DataFrame(jb_log_results)
print("--- KOCAELİ DOĞAL LOGARİTMALI SERİLER JARQUE-BERA NORMALLİK TESTİ ---")
print(df_jb_log.to_string(index=False))

# Sonuçları Excel dosyasına kaydetmek istersen:
# df_jb_log.to_excel('Kocaeli_Log_Jarque_Bera.xlsx', index=False)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2
from statsmodels.tsa.stattools import adfuller

# 1. Veriyi okuma ve log dönüşümü yapma
df = pd.read_excel('KOCAELI.xlsx')
numeric_cols = [col for col in df.columns if col != 'ZAMAN']

df_log = pd.DataFrame()
df_log['ZAMAN'] = df['ZAMAN']
for col in numeric_cols:
  df_log[f'ln_{col}'] = np.log(df[col])


# 2. Üç ayrı panel için ADF-Fisher Testi fonksiyonu
def run_panel_adffisher(column_list, panel_title):
  print(f'==================================================')
  print(f'PANEL GRUBU: {panel_title}')
  print(f'==================================================')

  results_data = []
  p_values = []

  for col in column_list:
    series = df_log[col].dropna()
    res = adfuller(series, regression='c', autolag='AIC')
    adf_stat = res[0]
    p_val = res[1]
    p_values.append(p_val)

    results_data.append({
        'İlçe / Seri': col,
        'ADF İstatistiği': adf_stat,
        'p-değeri': p_val,
        'Durağan mı? (p < 0.05)': 'Evet' if p_val < 0.05 else 'Hayır',
    })

  res_df = pd.DataFrame(results_data)
  display(res_df)

  # Fisher Chi-Square İstatistiği (Choi, 2001 / Maddala & Wu, 1999)
  valid_p = np.array(p_values)
  valid_p = np.clip(
      valid_p, 1e-15, 1.0
  )  # 0 p-değerleri için sayısal kararlılık

  fisher_stat = -2 * np.sum(np.log(valid_p))
  deg_freedom = 2 * len(valid_p)
  fisher_pval = 1 - chi2.cdf(fisher_stat, deg_freedom)

  print(f'--- {panel_title} PANEL ADF-FISHER ÖZETİ ---')
  print(f'Fisher Chi-Square İstatistiği : {fisher_stat:.4f}')
  print(f'Serbestlik Derecesi (df)       : {deg_freedom}')
  print(f'Fisher p-değeri                : {fisher_pval:.4f}')
  if fisher_pval < 0.05:
    print(
        'Sonuç: H0 reddedilir. Panel bütününde seriler DURAĞANDIR (I(0)).\n'
    )
  else:
    print(
        'Sonuç: H0 reddedilemez. Panel bütününde birim kök vardır, DURAĞAN'
        ' DEĞİLDİR.\n'
    )

  return fisher_stat, fisher_pval


# 3. Sütunları üç ayrı gruba (panele) ayırma
toplam_cols = [col for col in df_log.columns if 'TOPLAM' in col]
ipotek_cols = [col for col in df_log.columns if 'IPOTEK' in col]
ilk_cols = [col for col in df_log.columns if 'ILK' in col]

# 4. Üç ayrı paneli ayrı ayrı çalıştırma
stat_t, p_t = run_panel_adffisher(
    toplam_cols, '1. PANEL: TOPLAM KONUT SATIŞLARI'
)
stat_i, p_i = run_panel_adffisher(
    ipotek_cols, '2. PANEL: İPOTEKLİ KONUT SATIŞLARI'
)
stat_ilk, p_ilk = run_panel_adffisher(
    ilk_cols, '3. PANEL: İLK EL KONUT SATIŞLARI'
)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

# 1. Veri setini yüklüyoruz
file_path = 'KOCAELI.xlsx'
df = pd.read_excel(file_path, sheet_name='Sayfa1')

# Not: Verinin panel formatına (Long format) dönüştürülmesi gerekiyor.
# Geniş formattaki (Wide) veriyi long formata çevirelim:
districts = ['BAŞISKELE', 'ÇAYIROVA', 'DARICA', 'DERINCE', 'GEBZE', 'GOLCUK', 'IZMIT', 'KANDIRA', 'KARAMURSEL', 'KARTEPE', 'KORFEZ']

long_data = []
for index, row in df.iterrows():
    zaman = row['ZAMAN']
    for dist in districts:
        # Sütun adları eşleştirmesi
        toplam_val = row[f'{dist}-TOPLAM']
        ipotek_val = row[f'{dist}-IPOTEK']
        ilk_col = [c for c in df.columns if dist in c and 'ILK' in c][0]
        ilk_val = row[ilk_col]

        long_data.append({
            'ZAMAN': zaman,
            'ILCE': dist,
            'ln_TOPLAM': np.log(toplam_val),
            'ln_IPOTEK': np.log(ipotek_val),
            'ln_ILK': np.log(ilk_val)
        })

df_panel = pd.DataFrame(long_data)

# 2. Körfezray Güzergahı için Muamele (Treatment) Grubu Tanımı
# Örnek olarak Körfezray aksındaki ana ilçeleri treated ilan edelim (İzmit, Derince, Körfez, Kartepe)
treated_districts = ['IZMIT', 'DERINCE', 'KORFEZ', 'KARTEPE']
df_panel['Treated'] = df_panel['ILCE'].isin(treated_districts).astype(int)

# 3. Post (Etki Dönemi) Tanımı (Örn: 2020 ve sonrası proje yayılım dönemi olsun)
df_panel['Post'] = (df_panel['ZAMAN'] >= 2020).astype(int)

# 4. DiD Etkileşim Terimi
df_panel['DiD'] = df_panel['Treated'] * df_panel['Post']

print("=====================================================================")
print("          DIF-IN-DIFF (DID) İKİ YÖNLÜ SABİT ETKİLER MODELİ")
print("=====================================================================")

# Üç ayrı panel için DiD Regresyonu
categories = ['ln_TOPLAM', 'ln_IPOTEK', 'ln_ILK']

for cat in categories:
    print(f"\n--- Bağımlı Değişken: {cat} ---")
    formula = f"{cat} ~ DiD + C(ILCE) + C(ZAMAN)"

    # Panel regresyonu (OLS with Fixed Effects via Category Dummies)
    model = smf.ols(formula, data=df_panel).fit(cov_type='HC1') # Robust robust standart hatalar

    # Sadece DiD katsayısını ve istatistiklerini yazdıralım
    did_coef = model.params['DiD']
    did_pval = model.pvalues['DiD']
    did_se = model.bse['DiD']

    print(f"DiD Katsayısı (Beta) : {did_coef:.4f}")
    print(f"Standart Hata        : {did_se:.4f}")
    print(f"p-değeri             : {did_pval:.4f}")
    print(f"R-squared (Adj.)     : {model.rsquared_adj:.4f}")
```

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# 1. Veri setini yüklüyoruz
file_path = 'KOCAELI.xlsx'
df = pd.read_excel(file_path, sheet_name='Sayfa1')

districts = ['BAŞISKELE', 'ÇAYIROVA', 'DARICA', 'DERINCE', 'GEBZE', 'GOLCUK', 'IZMIT', 'KANDIRA', 'KARAMURSEL', 'KARTEPE', 'KORFEZ']

long_data = []
for index, row in df.iterrows():
    zaman = row['ZAMAN']
    for dist in districts:
        toplam_val = row[f'{dist}-TOPLAM']
        ipotek_val = row[f'{dist}-IPOTEK']
        ilk_col = [c for c in df.columns if dist in c and 'ILK' in c][0]
        ilk_val = row[ilk_col]

        long_data.append({
            'ZAMAN': zaman,
            'ILCE': dist,
            'ln_TOPLAM': np.log(toplam_val),
            'ln_IPOTEK': np.log(ipotek_val),
            'ln_ILK': np.log(ilk_val)
        })

df_panel = pd.DataFrame(long_data)

# 2. Gerçek Milat Tanımı (2023)
df_panel['Post'] = (df_panel['ZAMAN'] >= 2023).astype(int)

# Tedavi grubunun orijinal üyeleri
original_treated = ['IZMIT', 'DERINCE', 'KORFEZ', 'KARTEPE']

print("=====================================================================")
print("      JACKKNIFE (LEAVE-ONE-OUT) ANALİZİ SONUÇLARI")
print("=====================================================================")

# 3. Döngü: Her seferinde bir ilçeyi dışarıda bırakarak 4 iterasyon yapıyoruz
for excluded_district in original_treated:
    # Dışarıda bırakılan hariç, kalan 3 ilçe yeni tedavi grubu olur
    current_treated = [d for d in original_treated if d != excluded_district]

    # Geçici bir kopya üzerinden çalışma yapalım
    df_temp = df_panel.copy()
    df_temp['Treated'] = df_temp['ILCE'].isin(current_treated).astype(int)
    df_temp['DiD'] = df_temp['Treated'] * df_temp['Post']

    print(f"\n---------------------------------------------------------------------")
    print(f"DIŞARIDA BIRAKILAN İLÇE: {excluded_district} (Kalan Tedavi Grubu: {current_treated})")
    print(f"---------------------------------------------------------------------")

    categories = ['ln_TOPLAM', 'ln_IPOTEK', 'ln_ILK']
    for cat in categories:
        formula = f"{cat} ~ DiD + C(ILCE) + C(ZAMAN)"
        model = smf.ols(formula, data=df_temp).fit(cov_type='HC1')
        coef = model.params['DiD']
        pval = model.pvalues['DiD']
        print(f"Bağımlı Değişken: {cat:<10} | DiD Katsayısı: {coef:8.4f} | p-değeri: {pval:6.4f}")
```

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from arch import arch_model

# 1. Veri setini yüklüyoruz
file_path = 'KOCAELI.xlsx'
df = pd.read_excel(file_path, sheet_name='Sayfa1')

districts = ['BAŞISKELE', 'ÇAYIROVA', 'DARICA', 'DERINCE', 'GEBZE', 'GOLCUK', 'IZMIT', 'KANDIRA', 'KARAMURSEL', 'KARTEPE', 'KORFEZ']

long_data = []
for index, row in df.iterrows():
    zaman = row['ZAMAN']
    for dist in districts:
        toplam_val = row[f'{dist}-TOPLAM']
        ipotek_val = row[f'{dist}-IPOTEK']
        ilk_col = [c for c in df.columns if dist in c and 'ILK' in c][0]
        ilk_val = row[ilk_col]

        long_data.append({
            'ZAMAN': zaman,
            'ILCE': dist,
            'ln_TOPLAM': np.log(toplam_val),
            'ln_IPOTEK': np.log(ipotek_val),
            'ln_ILK': np.log(ilk_val)
        })

df_panel = pd.DataFrame(long_data)

categories = ['ln_TOPLAM', 'ln_IPOTEK', 'ln_ILK']

print("=====================================================================")
print("   ÜÇ AYRI PANEL İÇİN İLÇE BAZINDA GARCH(1,1) VOLATİLİTE KALICILIĞI")
print("=====================================================================")

for cat in categories:
    print(f"\n---------------------------------------------------------------------")
    print(f"BAĞIMLI DEĞİŞKEN: {cat}")
    print(f"---------------------------------------------------------------------")

    # İlgili kategori için panel OLS kalıntılarını alıyoruz
    formula = f"{cat} ~ C(ILCE) + C(ZAMAN)"
    baseline_model = smf.ols(formula, data=df_panel).fit()
    df_panel[f'Residuals_{cat}'] = baseline_model.resid

    for dist in districts:
        sub_df = df_panel[df_panel['ILCE'] == dist].sort_values('ZAMAN')
        res_series = sub_df[f'Residuals_{cat}'].dropna() * 100

        try:
            garch = arch_model(res_series, vol='Garch', p=1, q=1, rescale=False)
            garch_fitted = garch.fit(disp='off')
            alpha = garch_fitted.params['alpha[1]']
            beta = garch_fitted.params['beta[1]']
            persistence = alpha + beta
            print(f"İlçe: {dist:<12} | ARCH(alpha): {alpha:6.4f} | GARCH(beta): {beta:6.4f} | Kalıcılık (α+β): {persistence:6.4f}")
        except Exception as e:
            print(f"İlçe: {dist:<12} | Model yakınsamadı.")
```

```python
from arch import arch_model
from google.colab import files
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import statsmodels.formula.api as smf

# 0. Dosya Yükleme (Eğer yüklenmediyse)
print('Lütfen KOCAELI.xlsx dosyanızı seçin:')
uploaded = files.upload()
file_path = list(uploaded.keys())[0]

# 1. Veri setini yüklüyoruz
df = pd.read_excel(file_path, sheet_name='Sayfa1')

districts = [
    'BAŞISKELE',
    'ÇAYIROVA',
    'DARICA',
    'DERINCE',
    'GEBZE',
    'GOLCUK',
    'IZMIT',
    'KANDIRA',
    'KARAMURSEL',
    'KARTEPE',
    'KORFEZ',
]

long_data = []
for index, row in df.iterrows():
  zaman = row['ZAMAN']
  for dist in districts:
    toplam_val = row[f'{dist}-TOPLAM']
    ipotek_val = row[f'{dist}-IPOTEK']
    ilk_col = [c for c in df.columns if dist in c and 'ILK' in c][0]
    ilk_val = row[ilk_col]

    long_data.append({
        'ZAMAN': zaman,
        'ILCE': dist,
        'ln_TOPLAM': np.log(toplam_val),
        'ln_IPOTEK': np.log(ipotek_val),
        'ln_ILK': np.log(ilk_val),
    })

df_panel = pd.DataFrame(long_data)
categories = ['ln_TOPLAM', 'ln_IPOTEK', 'ln_ILK']

# 2. GARCH kalıcılık katsayılarını dinamik olarak toplama
garch_results_list = []

print('Running GARCH models and building persistence matrix...')
for cat in categories:
  formula = f'{cat} ~ C(ILCE) + C(ZAMAN)'
  baseline_model = smf.ols(formula, data=df_panel).fit()
  df_panel[f'Residuals_{cat}'] = baseline_model.resid

  for dist in districts:
    sub_df = df_panel[df_panel['ILCE'] == dist].sort_values('ZAMAN')
    res_series = sub_df[f'Residuals_{cat}'].dropna() * 100

    try:
      garch = arch_model(res_series, vol='Garch', p=1, q=1, rescale=False)
      garch_fitted = garch.fit(disp='off')
      alpha = garch_fitted.params['alpha[1]']
      beta = garch_fitted.params['beta[1]']
      persistence = alpha + beta
    except Exception as e:
      persistence = np.nan

    garch_results_list.append(
        {'District': dist, 'Category': cat, 'Persistence': persistence}
    )

# Sonuçları matrise dönüştürme ve sütun adlarını İngilizceleştirme
df_results = pd.DataFrame(garch_results_list)
df_heatmap = df_results.pivot(
    index='District', columns='Category', values='Persistence'
)
df_heatmap = df_heatmap.rename(
    columns={
        'ln_TOPLAM': 'Total Sales',
        'ln_IPOTEK': 'Mortgage Sales',
        'ln_ILK': 'First-hand Sales',
    }
)

# 3. SSCI/SCI Standartlarında İngilizce Isı Haritası (Heatmap) Görselleştirmesi
plt.figure(figsize=(9, 7), dpi=300)
sns.set_theme(style='white')

ax = sns.heatmap(
    df_heatmap,
    annot=True,
    cmap='YlOrRd',
    fmt='.4f',
    linewidths=0.8,
    linecolor='white',
    cbar_kws={'label': 'Volatility Persistence (Alpha + Beta)'},
    vmin=0.4,
    vmax=1.0,
)

plt.title(
    'Kocaeli Housing Market District-Level GARCH(1,1) Volatility Persistence'
    ' Heatmap',
    fontsize=11,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Market Segments', fontsize=11, fontweight='bold', labelpad=10)
plt.ylabel('Districts', fontsize=11, fontweight='bold', labelpad=10)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10, rotation=0)

plt.tight_layout()
plt.show()
```
